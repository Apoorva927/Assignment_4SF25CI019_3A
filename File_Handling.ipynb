1. Counting Words in a File

    Write a function that reads the contents of a file named `document.txt` and returns the number of words in the file.

```python
def count_txt_file(file_path):
    with open(file_path, 'r')as file:
        lines = file.readlines()
        word_count = sum(len(line.split()) for line in lines)
        return word_count
        
file_path = 'C:/Users/User/OneDrive/Desktop/html/Bootcamp/my_env/Assignments(Skill_Lab).ipynb/document.txt'
words = count_txt_file(file_path)
print(f'Number of Words: {words}')
```

2. Finding and Replacing Text

    Write a function that finds and replaces all occurrences of a given word in a file named `data.txt` with another word.

```python
def replace_word(file_path):
    with open(file_path, 'r')as file:
        text = file.read()

    old = input("Enter the word to replace: ")
    new = input("Enter new word: ")
    text = text.replace(old, new)

    with open(file_path, 'w')as file:
        file.write(text)
file_path = "C:/Users/User/OneDrive/Desktop/html/Bootcamp/my_env/Assignments(Skill_Lab).ipynb/data.txt"
replace_word(file_path)
```

3. Reading a File in Reverse

    Write a function that reads the contents of a file named `reverse.txt` and prints each line in reverse order.


```python
def count_txt_file(file_path):
    with open(file_path, 'r')as file:
        lines = file.readlines()
        for line in lines:
            print(line.strip()[::-1])
file_path = "C:/Users/User/OneDrive/Desktop/html/Bootcamp/my_env/Assignments(Skill_Lab).ipynb/reverse.txt"
replace_word(file_path)
```

4. Counting Lines, Words, and Characters

    Write a function that reads the contents of a file named `stats.txt` and returns the number of lines, words, and characters in the file.

```python
def count_txt_file(file_path):
    with open(file_path, 'r') as file:
        lines = file.readlines()
        line_count = len(lines)
        word_count = sum(len(line.split()) for line in lines)
        char_count = sum(len(line) for line in lines)
    return line_count, word_count, char_count

file_path = 'C:/Users/User/OneDrive/Desktop/html/Bootcamp/my_env/Assignments(Skill_Lab).ipynb/stats.txt'
lines, words, characters = count_txt_file(file_path)
print(f'Lines: {lines}, Words: {words}, Characters: {characters}')
```
